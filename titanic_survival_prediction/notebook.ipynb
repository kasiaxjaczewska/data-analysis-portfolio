# Analiza eksploracyjna Titanica

Notebook zawiera tylko EDA dla cech `age`, `sex` i `class` oraz zapisuje wykresy i crosstaby. Trening modeli znajduje się w `train.py`.

```python
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# EDA: wczytanie danych i zapis wykresów

df = pd.read_csv('data/titanic_clean.csv')
df.columns = df.columns.str.strip()

if 'survived' not in df.columns and 'Survived' in df.columns:
    df['survived'] = pd.to_numeric(df['Survived'], errors='coerce').fillna(0).astype(int)
else:
    df['survived'] = pd.to_numeric(df.get('survived'), errors='coerce').fillna(0).astype(int)

if 'age' not in df.columns:
    df['age'] = pd.to_numeric(df.get('Age'), errors='coerce')
else:
    df['age'] = pd.to_numeric(df['age'], errors='coerce')

if 'sex' not in df.columns:
    df['sex'] = df.get('Sex').astype(str).str.strip()
else:
    df['sex'] = df['sex'].astype(str).str.strip()

if 'class' not in df.columns:
    if 'Pclass' in df.columns:
        df['class'] = df['Pclass'].astype(str)
    else:
        df['class'] = df.get('Class').astype(str)
else:
    df['class'] = df['class'].astype(str)

print('Liczba wierszy i kolumn:', df.shape)
print(df[['age', 'sex', 'class', 'survived']].head())

# Rozkład wieku
plt.figure(figsize=(6, 4))
sns.histplot(df['age'].dropna(), bins=30, kde=True)
plt.title('Rozkład wieku')
plt.tight_layout()
plt.savefig('eda_age_distribution.png')
plt.close()

# Liczba obserwacji per płeć i klasa
plt.figure(figsize=(5, 3))
sns.countplot(data=df, x='sex')
plt.title('Liczba obserwacji per sex')
plt.tight_layout()
plt.savefig('eda_count_sex.png')
plt.close()

plt.figure(figsize=(5, 3))
sns.countplot(data=df, x='class')
plt.title('Liczba obserwacji per class')
plt.tight_layout()
plt.savefig('eda_count_class.png')
plt.close()

# Heatmapa braków
plt.figure(figsize=(6, 3))
sns.heatmap(df.isnull(), cbar=False)
plt.title('Heatmapa braków danych')
plt.tight_layout()
plt.savefig('eda_missing_heatmap.png')
plt.close()

# Crosstaby survived vs sex/class
ct_sex = pd.crosstab(df['sex'], df['survived'], normalize='index')
ct_class = pd.crosstab(df['class'], df['survived'], normalize='index')
ct_sex.to_csv('eda_crosstab_sex_survived.csv')
ct_class.to_csv('eda_crosstab_class_survived.csv')

print('Zapisano wykresy i crosstaby do plików EDA.')
```
